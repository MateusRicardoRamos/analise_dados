## 2. Vetor (1D)

Um **vetor** é uma sequência em **uma dimensão**: só tem comprimento.

Pense numa fila de notas: `[7.5, 8.0, 6.5, 9.0]`.

No NumPy isso vira um `ndarray` com `shape` `(n,)` — *n* elementos, uma linha só de números.

````python
notas = np.array([7.5, 8.0, 6.5, 9.0])

print(notas)
print("tipo:", type(notas))
print("ndim (quantas dimensões):", notas.ndim)
print("shape (formato):", notas.shape)
print("dtype (tipo dos números):", notas.dtype)
print("primeiro elemento:", notas[0])
print("média:", notas.mean())
````

````python
dimensao = notas.ndim
````

````python
duas_dimensoes = np.array([["nome", "nota"], [5, 6]])
````

````python
duas_dimensoes
````

````python
duas_dimensoes.shape
````

Operações em vetor valem **elemento a elemento** (não precisa de `for`):

````python
print("notas + 1 ->", notas + 1)
print("notas * 2 ->", notas * 2)

print("só as notas aprovadas ->", notas[notas >= 7])
````

````python
print("aprovados (>= 7) ->", notas >= 7)
````

## 3. Matriz (2D)

Uma **matriz** tem **duas dimensões**: linhas e colunas. `shape` vira `(linhas, colunas)`.

Exemplo: 3 alunos × 4 notas (provas).

````python
provas = np.array(
    [
        [7.0, 8.0, 6.5, 9.0],  # aluno 0
        [5.5, 6.0, 7.0, 6.5],  # aluno 1
        [9.0, 8.5, 9.5, 10.0],  # aluno 2
    ]
)

print(provas)
print("ndim:", provas.ndim)
print("shape:", provas.shape)  # (3 linhas, 4 colunas)
print("linhas:", provas.shape[0], "| colunas:", provas.shape[1])
````

Acesso em matriz: `matriz[linha, coluna]`.

- `provas[0]` → primeira **linha** (vetor das 4 notas do aluno 0)
- `provas[:, 0]` → primeira **coluna** (1ª prova de todo mundo)
- `:` significa “todos” naquela dimensão

````python
print("aluno 0 (linha):", provas[0])
print("prova 1 de todos (coluna):", provas[:, 1])
print("aluno 2, prova 3:", provas[2, 3])
print()
print("média por aluno (eixo 1 = colunas):", provas.mean(axis=1))
print("média por prova (eixo 0 = linhas):", provas.mean(axis=0))
````

### Vetor × matriz — quadro

| | Vetor | Matriz |
|--|--------|--------|
| Dimensões | 1 (`ndim=1`) | 2 (`ndim=2`) |
| `shape` | `(n,)` | `(linhas, colunas)` |
| Analogia | uma fila | uma grade / planilha só de números |
| Acesso | `v[i]` | `m[i, j]` |

O NumPy é ótimo para conta. Falta **nomear** as posições (“Ana”, “prova 2”). Isso a **Series** resolve no eixo 1D.

## 4. Series: um vetor com rótulos

`pd.Series` é um vetor (como o NumPy) **mais um índice**.

- `values` → os dados (por baixo, NumPy)
- `index` → o nome de cada posição
- `name` → o nome da série (opcional)

Se você não passar índice, o pandas usa `0, 1, 2, ...`.

````python
notas_serie = pd.Series([7.5, 8.0, 6.5, 9.0])
notas_serie
````

````python
print("values (NumPy por baixo):", notas_serie.values)
print("index:", notas_serie.index)
print("dtype:", notas_serie.dtype)
print("size:", notas_serie.size)
print("shape:", notas_serie.shape)
print("ndim:", notas_serie.ndim)
````

### 4.1 Series com índice nomeado

Agora cada nota tem um **rótulo** — bem mais legível que a posição 0, 1, 2.

````python
notas_alunos = pd.Series(
    [7.5, 8.0, 6.5, 9.0],
    index=["Ana", "Bruno", "Carla", "Diego"],
    name="nota",
)
notas_alunos
````

````python
notas_alunos
````

### 4.2 A partir de um dicionário

As **chaves** viram índice; os **valores** viram dados.

````python
temperaturas = pd.Series(
    {"BH": 22.0, "SP": 19.5, "RJ": 27.0, "POA": 16.0},
    name="celsius",
)
temperaturas
````

### 4.3 A partir de um vetor NumPy

A Series **embrulha** o vetor e ganha índice.

````python
vetor = np.array([10, 20, 30, 40])
serie_do_vetor = pd.Series(vetor, index=["a", "b", "c", "d"], name="valores")
serie_do_vetor
````

## 5. Acessar valores: `loc` e `iloc`

| Jeito | Olha para | Exemplo |
|-------|-----------|---------|
| `loc` | **rótulo** do índice | `notas_alunos.loc["Ana"]` |
| `iloc` | **posição** (0, 1, 2…) | `notas_alunos.iloc[0]` |
| `[]` | em geral o rótulo (pode confundir) | `notas_alunos["Ana"]` |

Na sala: use **`loc` / `iloc`**. Fica explícito se você quer nome ou posição.

````python
print("loc Ana:", notas_alunos.loc["Ana"])
print("iloc posição 0:", notas_alunos.iloc[0])
print()
print("fatiar por rótulo (inclui o fim):")
print(notas_alunos.loc["Ana":"Carla"])
print()
print("fatiar por posição (não inclui o fim):")
print(notas_alunos.iloc[0:2])
````

`loc` no fatiamento **inclui** o último rótulo. `iloc` segue a regra da lista Python: o fim fica de fora.

## 6. Operações e estatística

Como no vetor NumPy: conta elemento a elemento. O índice **viaja junto**.

````python
print(notas_alunos + 1)
print()
print("média:", notas_alunos.mean())
print("mediana:", notas_alunos.median())
print("mín / máx:", notas_alunos.min(), notas_alunos.max())
print("soma:", notas_alunos.sum())
print("desvio padrão:", notas_alunos.std())
````

````python
recuperacao = pd.Series(
    [0.5, 1.0, 1.5, 0.0],
    index=["Ana", "Bruno", "Carla", "Diego"],
)

# Soma pelo ÍNDICE (Ana com Ana, Bruno com Bruno), não pela ordem da lista
notas_alunos + recuperacao
````

Se os índices não baterem, o pandas coloca `NaN` (valor ausente) onde não houver par.

````python
extra = pd.Series([2.0, 1.0], index=["Ana", "Eva"])
notas_alunos + extra
````

## 7. Filtro (máscara booleana)

Uma comparação devolve uma Series de `True`/`False`. Usamos isso para **escolher linhas**.

````python
mascara = notas_alunos >= 7
print(mascara)
print()
print("quem passou:")
print(notas_alunos[mascara])
print()
print("quem ficou abaixo de 7:")
print(notas_alunos[notas_alunos < 7])
````

## 8. Valores ausentes (`NaN`)

Dado que falta aparece como `NaN`. Métodos úteis: `isna()`, `fillna()`, `dropna()`.

````python
medicoes = pd.Series([22.0, np.nan, 27.0, 16.0], index=["BH", "SP", "RJ", "POA"])
print(medicoes)
print()
print("onde falta:\n", medicoes.isna())
print("preencher com 0:\n", medicoes.fillna(0))
print("remover ausentes:\n", medicoes.dropna())
````

## 9. Lista × NumPy × Series

Mesmos números, três jeitos. Compare na prática:

````python
lista = [7.5, 8.0, 6.5, 9.0]
vetor = np.array(lista)
serie = pd.Series(lista, index=["Ana", "Bruno", "Carla", "Diego"], name="nota")

print("lista[0] =", lista[0], "| type:", type(lista))
print("vetor[0] =", vetor[0], "| shape:", vetor.shape)
print("serie['Ana'] =", serie.loc["Ana"], "| index:", list(serie.index))
````

## Recapitulando

| Conceito | Biblioteca | Dimensão | Tem rótulo? |
|----------|------------|----------|-------------|
| vetor | NumPy `array` | 1 | não (só posição) |
| matriz | NumPy `array` | 2 | não |
| Series | pandas | 1 | **sim** (`index`) |

- `shape` descreve o tamanho: `(n,)` vetor, `(linhas, colunas)` matriz.
- Series = valores + índice. Use `loc` (nome) e `iloc` (posição).

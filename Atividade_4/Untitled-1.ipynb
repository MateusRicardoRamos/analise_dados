```python
populacao_estado = (
    df.groupby("UF")[[
        "POPULACAO_2010_COMPATIBILIZADA",
        "POPULACAO_2022"
    ]]
    .sum()
    .reset_index()
)
```

```python
print(populacao_estado)
```

```python
populacao_estado["CRESCIMENTO_POPULACAO"] = (
    populacao_estado["POPULACAO_2022"]
    - populacao_estado["POPULACAO_2010_COMPATIBILIZADA"]
)
```

```python
populacao_estado = populacao_estado.sort_values(
    "CRESCIMENTO_POPULACAO",
    ascending=False
)
```

```python
populacao_estado.to_csv(
    "populacao_por_estado.csv",
    sep=";",
    index=False
)
```

```python
populacao_municipio = df[[
    "UF",
    "COD_UF",
    "COD_MUNIC",
    "NOME_MUNICIPIO",
    "POPULACAO_2010_COMPATIBILIZADA",
    "POPULACAO_2022"
]].copy()
```

```python
populacao_municipio["CRESCIMENTO_POPULACAO"] = (
    populacao_municipio["POPULACAO_2022"]
    - populacao_municipio["POPULACAO_2010_COMPATIBILIZADA"]
)
```

```python
populacao_municipio = populacao_municipio.sort_values(
    "CRESCIMENTO_POPULACAO",
    ascending=False
)
```
